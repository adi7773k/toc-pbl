```python
def palindrome_algorithm_to_tm():
    states = []
    transitions = []
    state_counter = 0

    def new_state():
        nonlocal state_counter
        state = f"q{state_counter}"
        state_counter += 1
        states.append(state)
        return state

    blank = '_'
    accept_state = 'ACCEPT'
    reject_state = 'REJECT'

    tape_alphabet = ['0', '1', 'X', 'Y', blank]
    input_alphabet = ['0', '1']

    # Define states
    q_start = new_state()
    q_check_end = new_state()
    q_scan_right_0 = new_state()
    q_scan_right_1 = new_state()
    q_back_to_start = new_state()
    q_check_done = new_state()

    # Step 1: From start, check leftmost unmarked symbol
    transitions += [
        (q_start, 'X', 'X', 'R', q_start),
        (q_start, 'Y', 'Y', 'R', q_start),
        (q_start, '0', 'X', 'R', q_scan_right_0),
        (q_start, '1', 'Y', 'R', q_scan_right_1),
        (q_start, blank, blank, 'R', q_check_done),
    ]

    # Step 2: Scan right to find matching end symbol for 0
    transitions += [
        (q_scan_right_0, '0', '0', 'R', q_scan_right_0),
        (q_scan_right_0, '1', '1', 'R', q_scan_right_0),
        (q_scan_right_0, 'X', 'X', 'R', q_scan_right_0),
        (q_scan_right_0, 'Y', 'Y', 'R', q_scan_right_0),
        (q_scan_right_0, blank, blank, 'L', q_check_end),
    ]

    # Step 3: At end, verify matching symbol
    transitions += [
        (q_check_end, '0', 'X', 'L', q_back_to_start),  # match for 0
        (q_check_end, '1', '1', 'L', reject_state),     # mismatch
        (q_check_end, 'X', 'X', 'L', q_check_end),
        (q_check_end, 'Y', 'Y', 'L', q_check_end),
    ]

    # Step 4: Scan right to find matching end symbol for 1
    transitions += [
        (q_scan_right_1, '0', '0', 'R', q_scan_right_1),
        (q_scan_right_1, '1', '1', 'R', q_scan_right_1),
        (q_scan_right_1, 'X', 'X', 'R', q_scan_right_1),
        (q_scan_right_1, 'Y', 'Y', 'R', q_scan_right_1),
        (q_scan_right_1, blank, blank, 'L', q_check_end),
    ]

    # Step 5: At end, verify matching symbol
    transitions += [
        (q_check_end, '1', 'Y', 'L', q_back_to_start),  # match for 1
        (q_check_end, '0', '0', 'L', reject_state),     # mismatch
    ]

    # Step 6: Return to beginning
    transitions += [
        (q_back_to_start, '0', '0', 'L', q_back_to_start),
        (q_back_to_start, '1', '1', 'L', q_back_to_start),
        (q_back_to_start, 'X', 'X', 'L', q_back_to_start),
        (q_back_to_start, 'Y', 'Y', 'L', q_back_to_start),
        (q_back_to_start, blank, blank, 'R', q_start),
    ]

    # Step 7: Final check
    transitions += [
        (q_check_done, 'X', 'X', 'R', q_check_done),
        (q_check_done, 'Y', 'Y', 'R', q_check_done),
        (q_check_done, blank, blank, 'R', accept_state),
        (q_check_done, '0', '0', 'R', reject_state),
        (q_check_done, '1', '1', 'R', reject_state),
    ]

    return {
        "states": states + [accept_state, reject_state],
        "alphabet": input_alphabet,
        "tape_alphabet": tape_alphabet,
        "start_state": q_start,
        "accept_state": accept_state,
        "reject_state": reject_state,
        "transitions": transitions,
    }


def print_tm(tm):
    print("States:", tm["states"])
    print("Start State:", tm["start_state"])
    print("Accept State:", tm["accept_state"])
    print("Reject State:", tm["reject_state"])
    print("\nTransitions:")
    for (cur, read, write, move, nxt) in tm["transitions"]:
        print(f"δ({cur}, '{read}') = ('{write}', {move}, {nxt})")


# Run it
tm = palindrome_algorithm_to_tm()
print_tm(tm)
```
